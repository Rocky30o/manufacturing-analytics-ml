```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
comparison = pd.DataFrame(
    {
        "Model": [
            "Random Forest Multiclass",
            "XGBoost Multiclass",
            "XGBoost Binary"
        ],

        "Accuracy": [
            0.9915,
            0.9765,
            0.9785
        ],

        "Macro_F1": [
            0.6406,
            0.6468,
            0.8700
        ]
    }
)

comparison
```

```python
plt.figure(
    figsize=(8,5)
)

sns.barplot(
    data=comparison,
    x="Model",
    y="Accuracy"
)

plt.title(
    "Model Accuracy Comparison"
)

plt.xticks(
    rotation=20
)

plt.show()
```

```python
plt.figure(
    figsize=(8,5)
)

sns.barplot(
    data=comparison,
    x="Model",
    y="Macro_F1"
)

plt.title(
    "Macro F1 Comparison"
)

plt.xticks(
    rotation=20
)

plt.show()
```
